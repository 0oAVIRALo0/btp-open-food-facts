```python
import pickle
import pandas as pd

# Load the trained model
model_path = 'nutrients_7.pkl'
with open(model_path, 'rb') as file:
    model = pickle.load(file)
input_columns = [
    'fat_100g',
    'carbohydrates_100g',
    'proteins_100g',
    'sodium_100g',
    'sugars_100g',
    'saturated-fat_100g',
    'energy-kcal_100g'
]
input_data = df_selected[input_columns]
input_data.columns = [
    'Total Fat',             # Renamed from 'fat_100g'
    'Carbohydrate',          # Renamed from 'carbohydrates_100g'
    'Protein',               # Renamed from 'proteins_100g'
    'Sodium',                # Renamed from 'sodium_100g'
    'Sugars, total',         # Renamed from 'sugars_100g'
    'Fatty acids, total saturated',  # Renamed from 'saturated-fat_100g'
    'Energy'                 # Renamed from 'energy-kcal_100g'
]
predictions = model.predict(input_data)

# Add predictions as a new column
df_selected['predicted'] = predictions

# Save the updated DataFrame to a new CSV file
output_path = 'OFF_DATASET_selected_columns_extended_with_predictions.csv'
df_selected.to_csv(output_path, index=False)
print(f"Predictions added and saved to {output_path}")
```

```python
# Find the min, max, mean and median values for each column

numeric_columns = df.select_dtypes(include=["float64", "int64"]).columns

for column in numeric_columns:
    min_value = df[column].min()
    max_value = df[column].max()
    mean_value = df[column].mean()
    median_value = df[column].median()
    print(f"{column}: min={min_value}, max={max_value}, mean={mean_value}, median={median_value}")
```
